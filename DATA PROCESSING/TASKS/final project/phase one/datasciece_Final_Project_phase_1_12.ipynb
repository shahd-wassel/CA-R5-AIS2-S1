###  To fix this issue You will fill these values with ......median..... for numeric and .mode.....for categorical

```python
# trying to fill missing values in my datset using for loop
cat_cols = ["start_station_id", "start_station_name", "end_station_id", "end_station_name", "bike_id", "user_type", "bike_share_for_all_trip"]
df[cat_cols] = df[cat_cols].astype("category")

num_cols = ["duration_sec", "start_station_latitude", "start_station_longitude", "end_station_latitude", "end_station_longitude", "member_birth_year"]


for col in num_cols:
    median = df[col].median()
    df[col] = df[col].fillna(median)

for col in cat_cols:
    mode = df[col].mode().iloc[0]
    df[col] = df[col].fillna(mode)

df['member_gender'] = df['member_gender'].astype('category')

```

```python
#check the Nulls
df.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
##  still got a missing value in the member_gender, so let's explore this column more

print(df.member_gender.head(15))
```

```python
# fill  dataset with missing values.

df['member_gender'] = df['member_gender'].fillna('Other')
```

## limitation
example
- We got missing data which we fix some of them .

-start_time/end_time were corrupted (only minutes:seconds remained, date/hour lost) — likely from an Excel formatting issue — so no real time-of-day or day-of-week analysis was possible.
