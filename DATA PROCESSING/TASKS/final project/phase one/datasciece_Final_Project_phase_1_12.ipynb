

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with ......median..... for numeric and .mode.....for categorical

```python
# trying to fill missing values in my datset using for loop
cat_cols = ["start_station_id", "start_station_name", "end_station_id", "end_station_name", "bike_id", "user_type", "bike_share_for_all_trip"]
df[cat_cols] = df[cat_cols].astype("category")

num_cols = ["duration_sec", "start_station_latitude", "start_station_longitude", "end_station_latitude", "end_station_longitude", "member_birth_year"]


for col in num_cols:
    median = df[col].median()
    df[col] = df[col].fillna(median)

for col in cat_cols:
    mode = df[col].mode().iloc[0]
    df[col] = df[col].fillna(mode)

df['member_gender'] = df['member_gender'].astype('category')

```

```python
#check the Nulls
df.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
##  still got a missing value in the member_gender, so let's explore this column more

# print(df.member_gender.head(15))
```

```python
# fill  dataset with missing values.

df['member_gender'] = df['member_gender'].fillna('Other')
```

```python
## check my result above
print(df.isnull().sum())
df['member_gender'].value_counts()
```

```python
## I will import seaborn library to support me in making visualizations

import seaborn as sns
 plt.figure(figsize = (8,5))

 plot = sns.countplot(data=df, x='user_type')

 plt.xlabel('User type')
 plt.ylabel('count')
 plt.title("Represent the user's distribution")

 plt.show()
 df['user_type'].value_counts(normalize=True) * 100
```

# <a id="3"> 3 ) What is the distribution for users' ages? </a>

```python
plt.figure(figsize=(8,5))
plt.hist(data=df, x='age', bins=30)
plt.xlabel('Age')
plt.ylabel('Count')
plt.title("Distribution of User Age")
plt.show()
plt.figure(figsize=(9,2))

sns.kdeplot(df['age'])
plt.title("age hist Graph")

plt.show()
```

> ### <mark>Your analsys and exploration here </mark>
The age distribution is right-skewed, with most users between 20 and 40 years old. There's a long tail extending toward much older ages, with a small number of extreme outliers up to ~140 years — clearly impossible ages likely caused by incorrect birth years entered in the source data.
