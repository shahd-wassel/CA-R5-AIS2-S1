11. Find max and min values in a set

```python
my_set = {10, 3, 25, 1, 7}

# TODO: find max and min manually
# your code here
numbers = list(my_set)
maximum = numbers[0]
minimum = numbers[0]

for number in numbers:
    if number > maximum:
        maximum = number
    if number < minimum:
        minimum = number

print("Max:", maximum)
print("Min:", minimum)
# Hint: Convert to a list or loop through items.
```
