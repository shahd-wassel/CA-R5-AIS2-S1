```python
class BankAccount:
    """"class representing a bank account with basic operations like deposit, withdraw, and check balance.
    Attributes:
        balance (float): The current balance of the account.
    """

    def __init__(self, initial_balance=0):
        """initializes the BankAccount with an initial balance.
        args:
            initial_balance (float): The starting balance of the account. Defaults to 0.
        """
        self.balance = initial_balance


    
 
    def deposit(self, amount):
        """Deposits a specified amount into the account."""
        self.balance += amount
        print(f'Deposited {amount}. New balance: {self.balance}')

    def withdraw(self, amount):
        """withdraws a specified amount from the account """
        if amount > self.balance:
            print('Insufficient funds')
        else:
            self.balance -= amount
            print(f'Withdrew {amount}. New balance: {self.balance}')

    def check_balance(self):
        """checks balance """
        print(f'Current balance: {self.balance}')


        return self.deposit, self.withdraw, self.check_balance

# Usage
account = BankAccount(100)
deposit, withdraw, check_balance = account.deposit, account.withdraw, account.check_balance
deposit(50)
withdraw(30)
check_balance()
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    """
    Abstract Animal class that defines common behavior for animals.

    Methods:
        make_sound(): Abstract method that must be implemented by subclasses.
        describe(): Prints a general description of the animal.
    """

    @abstractmethod
    def make_sound(self):
        """
        Returns the sound made by the animal.

        Returns:
            str: The animal's sound.
        """
        pass

    def describe(self):
        """
        Prints a general description of the animal.
        """
        print("This is an animal.")


class Dog(Animal):
    """
    Dog class that inherits from Animal.
    """

    def make_sound(self):
        """
        Returns the sound made by a dog.

        Returns:
            str: 'Woof!'
        """
        return "Woof!"


class Cat(Animal):
    """
    Cat class that inherits from Animal.
    """

    def make_sound(self):
        """
        Returns the sound made by a cat.

        Returns:
            str: 'Meow!'
        """
        return "Meow!"


class Cow(Animal):
    """
    Cow class that inherits from Animal.
    """

    def make_sound(self):
        """
        Returns the sound made by a cow.

        Returns:
            str: 'Moo!'
        """
        return "Moo!"

dog = Dog()
print("Dog:")
dog.describe()
print("Sound:", dog.make_sound())

print("\nCat:")
cat = Cat()
cat.describe()
print("Sound:", cat.make_sound())

print("\nCow:")
cow = Cow()
cow.describe()
print("Sound:", cow.make_sound())
```

```python
class TextFileReader:
    """
    TextFileReader class handles reading a text file,
    counting lines, words, and characters, and displaying its content.

    Attributes:
        file_path (str): The path of the text file.
        content (str): The content of the text file.
    """

    def __init__(self, file_path):
        """
        Initializes the TextFileReader object.

        Args:
            file_path (str): The path of the text file.
        """
        self.file_path = file_path
        self.content = ""

    def read_file(self):
        """
        Opens the file and reads its content.

        Returns:
            str: The content of the file.
        """
        with open(self.file_path, "r") as file:
            self.content = file.read()

        return self.content

    def count_lines(self):
        """
        Counts the number of lines in the file.

        Returns:
            int: The number of lines.
        """
        return len(self.content.splitlines())

    def count_words(self):
        """
        Counts the total number of words in the file.

        Returns:
            int: The number of words.
        """
        return len(self.content.split())

    def count_characters(self):
        """
        Counts the total number of characters in the file.

        Returns:
            int: The number of characters.
        """
        return len(self.content)

    def display_content(self):
        """
        Displays the content of the file.
        """
        print(self.content)

file_reader = TextFileReader("example.txt")
file_reader.read_file()
print("File Content:")
file_reader.display_content()
print("Number of lines:", file_reader.count_lines())
print("Number of words:", file_reader.count_words())
print("Number of characters:", file_reader.count_characters())
```

```python
def count_words(words):
     """
        Counts the occurrences of each unique word in a list of words.
    
        Args:
            words (list): A list of words. 
        """ 
     unique_words = set(words)

     counts = {word: 0 for word in unique_words}
 
     for word in words:
        counts[word] += 1

     return counts


words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]

print(count_words(words))
```

```python
def read_txt_file(file_path):
    """Reads the contents of a text file and returns it as a string."""
    try:
        with open(file_path, "r", encoding="utf-8") as file:
            content = file.read()
        return content

    except FileNotFoundError:
        return f"Error: The file '{file_path}' was not found."

    except IOError:
        return "Error: An error occurred while reading the file."


class UserExtractor:
    def __init__(self, file_path):
        """
        Initializes the UserExtractor object.

        Args:
            file_path (str): The path to the text file.
        """
        self.file_path = file_path
        self.usernames = {}

    def extract_usernames(self):
        """Extracts usernames from the text file and stores them in a dictionary."""

        # Read the file using the external function
        content = read_txt_file(self.file_path)

        # Check if there was an error while reading the file
        if content.startswith("Error:"):
            return content

        # Process each line
        for line in content.splitlines():
            if ":" in line:
                username, password = line.split(":", 1)

                # Store only unique usernames
                self.usernames[username] = password

        return self.usernames


# Usage example
if __name__ == "__main__":
    file_path = "test.txt"

    # Create an object
    user_extractor = UserExtractor(file_path)

    # Extract usernames
    result = user_extractor.extract_usernames()

    # Print the result
    print(result)
```
