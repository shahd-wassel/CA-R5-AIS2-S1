Write a function that inputs a number and prints the multiplication table of that
number

```python
def multiplication_table(number):
    for i in range(1, 11):
        print(number, "x", i, "=", number * i)

number = int(input("Enter a number: "))
multiplication_table(number)
```

Write a program to print twin primes less than 1000. If two
consecutive odd numbers are both prime then they are known as
twin primes

```python
def is_prime(i):
    if i < 2:
        return False

    for t in range(2, i):
        if i % t == 0:
            return False
    return True 

def twin_primes():
    for i in range(3, 1000, 2):
        if is_prime(i) and is_prime(i + 2):
            print(i, i + 2)


twin_primes()
```

3. 3. Write a program to find out the prime factors of a number.
Example: prime factors of 56 - 2, 2, 2, 7

```python
def prime_factors(n):
    factors = []

    divisor = 2

    while n > 1:
        while n % divisor == 0:
            factors.append(divisor)
            n = n / divisor
            

        divisor += 1

    return factors


num = int(input("Enter a number: "))

print("Prime factors:", prime_factors(num))
```

4. Write a function that converts a decimal number to binary 
number

```python
def dectobin(n):
    binary = ""

    while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n = n // 2

    return binary


num = int(input("Enter a decimal number: "))
print("Binary number:", dectobin(num))
```

5. A number is called perfect if the sum of proper divisors of that
number is equal to the number. For example 28 is perfect
number, since 1+2+4+7+14=28. Write a program to print all the
perfect numbers in a given range

```python
def perfectnums(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return total == n


start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

print("Perfect numbers are:")

for num in range(start, end + 1):
    if perfectnums(num):
        print(num)
```
