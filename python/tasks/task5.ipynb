```python
class MultiplicationTable:
    '''
    MultiplicationTable class handles the operation of
    displaying the multiplication table of a number.

    Attributes:
        number (int): The number whose multiplication table will be displayed.
    '''

    def __init__(self, number):
        """
        Initializes the MultiplicationTable object.

        Args:
            number (int): The number for the multiplication table.
        """
        self.number = number

    def display_table(self):
        """
        Displays the multiplication table from 1 to 10.

        Returns:
            None
        """
        for i in range(1, 11):
            print(self.number, "x", i, "=", self.number * i)

number = int(input("Enter a number: "))
table = MultiplicationTable(number)
table.display_table()
```

```python
class primefactor:
     """
        primefactor class handles the operation of
        finding the prime factors of a number.
    
        Attributes:
            number (int): The number whose prime factors will be found.
        """
     def __init__(self, number):
        """
                Initializes the primefactor object.
        
                Args:
                    number (int): The number for which to find prime factors.
       """
        self.number = number
        self.number = number

     def prime_factors(self):
       """
              Finds the prime factors of the number.
      
              Returns:
                  list: A list of prime factors.
              """
       factors = []

       divisor = 2

       while self.number > 1:
        while self.number % divisor == 0:
            factors.append(divisor)
            self.number = self.number / divisor
       divisor += 1

       return factors
    
num = int(input("Enter a number: "))
pf = primefactor(num)
print("Prime factors:", pf.prime_factors())
```

```python
class DecimalToBinary:
    """
    DecimalToBinary class handles the conversion of a decimal number
    into its binary representation.

    Attributes:
        number (int): The decimal number to be converted.
    """

    def __init__(self, number):
        """
        Initializes the DecimalToBinary object.

        Args:
            number (int): The decimal number to convert.
        """
        self.number = number

    def convert(self):
        """
        Converts the decimal number to binary.

        Returns:
            str: The binary representation of the number.
        """
        binary = ""

        n = self.number

        while n > 0:
            remainder = n % 2
            binary = str(remainder) + binary
            n = n // 2

        return binary
    
num = int(input("Enter a decimal number: "))
obj = DecimalToBinary(num)
print("Binary number:", obj.convert())
```

```python
class PerfectNumbers:
    """
    PerfectNumbers class handles the operation of finding
    perfect numbers within a given range.

    Attributes:
        start (int): The starting number of the range.
        end (int): The ending number of the range.
    """

    def __init__(self, start, end):
        """
        Initializes the PerfectNumbers object.

        Args:
            start (int): The starting number.
            end (int): The ending number.
        """
        self.start = start
        self.end = end

    def is_perfect(self, n):
        """
        Checks whether a number is a perfect number.

        Args:
            n (int): The number to check.

        Returns:
            bool: True if the number is perfect, otherwise False.
        """
        total = 0

        for i in range(1, n):
            if n % i == 0:
                total += i

        return total == n

    def find_perfect_numbers(self):
        """
        Finds and returns all perfect numbers in the given range.

        Returns:
            list: A list containing the perfect numbers.
        """
        perfect_numbers = []

        for num in range(self.start, self.end + 1):
            if self.is_perfect(num):
                perfect_numbers.append(num)

        return perfect_numbers

start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))
obj = PerfectNumbers(start, end)
print("Perfect numbers are:")

for num in obj.find_perfect_numbers():
    print(num)
```
