```python
email = input("Enter your email address: ")
if email.count("@") != 1:
    print("Invalid email")
else:
    at_position = email.find("@")
    dot_position = email.find(".", at_position)

    if dot_position == -1:
        print("Invalid email")
    else:
        print("Valid email")
        username = email.split("@")[0]
        print("Username:", username)
        domain = email.split("@")[1].rsplit(".", 1)[0]
        print("Domain:", domain)
        if email.endswith(".com"):
            print("Commercial Domain")
        elif email.endswith(".edu"):
            print("Educational Domain")
        else:
            print("Other Domain")
```
